# 在空间中解析的单细胞数据

(空间-介绍-关键-外卖-1)=
## 动机

批量和单细胞基因组技术使我们能够表征和理解细胞身份及其对基因组规模的依赖性。到目前为止，本工作中描述的所有单细胞技术都表征了从感兴趣的样品中去除的解离细胞。然而，通过消除细胞及其分子的空间背景，人们也失去了空间背景，而空间背景是许多生物学问题的重要组成部分。空间解析基因组学通过测量基因组规模的组学并保留空间信息来解决这个问题。 

## 空间剖面测量概述

空间解析的基因组学可以通过量化转录组、蛋白质组或染色质可及性的各种技术来测量。然而，这些技术在规模、分辨率、灵敏度、复用性和适用性方面各不相同。由于空间分辨基因组学的前景正在快速发展，并且预计在未来几年内将发生巨大变化，因此我们根据其捕获分辨率仅介绍三组空间组学技术。可以通过本介绍之后链接的评论和论文获得更多信息。 {cite}`moffitt_emerging_2022`{cite}`walker_deciphering_2022`{cite}`yue_guidebook_2023`

从广义上讲，人们可以将空间分辨基因组学区分为以多细胞、单细胞和亚细胞分辨率进行测量的技术。我们将很快激发每个规模并强调它们的优势和挑战，并向读者介绍属于各自类别的一些技术。 



### 多单元分辨率
以多细胞分辨率获得的空间组学数据通常捕获多个细胞之间的组学测量。因此，每个数据点包含来自不同数量的细胞以及可能不同的细胞类型的信息。多细胞分辨率数据可以用反卷积方法分解以获得每个点不同细胞或细胞类型的比例。

多细胞分辨率数据通常以不同的分辨率捕获转录组范围的基因表达谱。基于点的技术获得的分辨率在 55um (Visium) 到 10 um (slide-seq) 之间变化。 10x Genomics 提供的 Visium 是一项广为人知、可商用且成功的技术。我们将在教程中展示如何分析和反卷积 Visium 数据。



### 单细胞分辨率
以单细胞分辨率获得的空间组学数据可以直接捕获准确位置的单细胞，也可以捕获单细胞尺度上的点。基于点的方法的示例有 HDST、slide-seqV2 或stereo-seq。这些方法捕获整个转录组，但捕获效率仍然较低。 

靶向方法提供了一种在细胞精确位置进行测量的替代方法。常见的例子有 MERFISH、seqFISH+、IMC 或多重 IHC（例如 cyCIF 和 CODEX）。这些技术通常很昂贵，并且只能测量有限的特征空间。这些方法不捕获预定义位置或网格处的点，而是测量单个转录物或细胞位置。



### 亚细胞分辨率
亚细胞分辨率的空间组学数据捕获单个 RNA 分子的位置。这些位置可以通过单分子成像或通过斑点尺寸小于单细胞的空间条形码来捕获。通过对亚细胞数据执行细胞分割，可以获得单细胞分辨率数据，其中表达被聚合为细胞级测量，然后可以在空间感知分析管道中进行处理。我们将展示如何对 MERFISH 数据集执行此分析。
