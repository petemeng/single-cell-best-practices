# 单细胞 RNA 测序

本章简要介绍了最广泛使用的单细胞核糖核酸 ({term}`RNA`) 测序分析和相关的基本分子生物学概念。
{term}`Multimodal <Modalities>`或空间分析不在此处介绍，但在相应的高级章节中介绍。
所有 {term}`sequencing`检测都有各自的优点和局限性，数据分析师必须了解这些优点和局限性，以了解数据中可能存在的偏差。

## 生命的基石

正如我们所知，生命是区分生物与死亡或无生命实体的特征。
生命一词的大多数定义都有一个共同的实体 - {term}`cells <Cell>`。
细胞形成开放系统，维持体内平衡、新陈代谢、生长、适应环境、繁殖、对刺激做出反应并自我组织。
因此，细胞是生命的基本组成部分，由英国科学家Robert Hooke于1665年首次发现。
胡克用非常简陋的显微镜研究了一小片软木塞，令他惊讶的是，他注意到这片软木片看起来像蜂窝。
他将这些微小的单位命名为“细胞”。

<img src="../_static/images/scrna_seq/hooke_cork.jpg" alt="Robert Hook cell" class="bg-primary mb-1" width="100%">

Robert Hooke 绘制的软木细胞图。图片来自 Micrographia。
:::

1839年，Matthias Jakob Schleiden和Theodor Schwann首次描述了Cell Theory，指出所有生物体都是由细胞组成的。
自 Cell Theory 的早期定义以来，研究人员发现所有细胞都具有几乎相同的化学成分，并表现出动态信息流，以脱氧核糖核酸 ({term}`DNA`) 的形式将遗传密码从一个细胞传递到另一个细胞。
存在两种一般类型的细胞：真核生物和原核生物。
真核细胞含有细胞核，核膜包裹着染色体；而原核细胞只有一个类核区，没有细胞核。
细胞核承载着细胞的基因组 DNA，这就是它们被称为真核生物的原因：_Nucleus_ 在拉丁语中是内核或种子的意思。
DNA 复制机制读取细胞核中 DNA 中存储的遗传信息来复制自身并保持生命周期的持续。
真核生物 DNA 被分成几个称为染色体的线性束，它们在核分裂过程中被微管纺锤体分开。
了解 DNA 中隐藏的遗传信息是了解许多进化和疾病相关过程的关键。

测序是破译 DNA 核苷酸顺序的过程。
它主要用于揭示特定 DNA 片段、完整基因组甚至复杂微生物组所携带的遗传信息。
DNA 测序使研究人员能够识别基因的位置、功能和调控。例如，它揭示了开放阅读框 (ORF)、起始密码子和终止密码子之间的蛋白质编码序列或指示 {term}`promoter`区域的 {term}`CpG`岛等遗传特征。
另一个广泛的应用领域是进化分析，其中比较来自不同生物体的同源 DNA 序列。
DNA 测序还可以应用于突变与疾病之间的关联，有时甚至可以应用于抗病性，这被认为是最有价值的应用之一。

## 测序简史

### 第一代测序

尽管 DNA 已于 1869 年由 Friedrich Miescher 首次分离出来，但科学界花了 100 多年的时间才开发出高通量测序技术。
1953 年，Watson、Crick 和 Franklin 发现了 DNA 的结构，1965 年，Robert Holley 对第一个 tRNA 进行了测序。
七年后，即 1972 年，Walter Fiers 首次使用 RNA 酶对完整基因（噬菌体 MS2 的外壳蛋白）进行测序，消化病毒 RNA，分离寡核苷酸，最后通过电泳和色谱法将其分离{cite}`Jou1972`。
与此同时，Frederick Sanger 使用放射性标记、部分消化的片段开发了一种 DNA 测序方法，称为“链终止法”，通常称为“Sanger Sequencing”。
Although Sanger Sequencing 至今仍在使用，但它存在一些缺点，包括缺乏自动化和耗时。
1987 年，Leroy Hood 和 Michael Hunkapiller 开发了 ABI 370，这是一种自动化 Sanger Sequencing 过程的仪器。
其最重要的创新成就是用荧光染料而不是放射性分子自动标记 DNA 片段。
这一变化不仅使该方法执行起来更安全，而且允许计算机分析获取的数据{cite}`Hood1987`。

```{dropdown} <i class="fa-solid fa-thumbs-up"></i>   优势
- Sanger sequencing is simple and affordable.
- If done correctly, the error rate is very low (<0.001%).
```

```{dropdown} <i class="fa-solid fa-thumbs-down"></i></i>   局限性
- Sanger methods can only sequence short pieces of DNA of about 300 to 1000 base pairs.
- The quality of a Sanger sequence is often poor in the first 15 to 40 bases because this is where the primers bind.
- Sequencing degrades after 700 to 900 bases.
- If the sequenced DNA fragment has been cloned, some of the cloning vector sequence (a DNA carrier for copying, storing, and amplifying genes) may find its way into the final sequence.
- Sanger sequencing is more expensive than second or third-generation sequencing per sequenced base.
```

### 二代测序

九年后的 1996 年，Mostafa Ronaghi、Mathias Uhlen 和 Pşl Nyŕen 通过开发焦磷酸测序彻底改变了 DNA 测序，标志着第二代测序的开始。
第二代测序，也称为下一代测序（NGS），主要是通过实验室的进一步自动化、计算机的使用和反应的小型化而成为可能。
焦磷酸测序测量测序过程中焦磷酸合成产生的发光。
该过程通常也称为“合成测序”。
两年后，Shankar Balasubramanian 和 David Klenerman 在 Solexa 公司开发并改造了合成测序流程，以实现一种利用荧光染料的新方法。
Solexa 的技术也构成了 Illumina 测序仪的基础，该测序仪在当今市场上占据主导地位。
The Roche 454 测序仪于 2005 年开发，是第一台在单一自动化机器中完全自动化焦磷酸测序过程的测序仪。
Life Technologies 推出了其他几个平台，包括 2007 年的 SOLiD（“连接测序”系统）和 2011 年的 Ion Torrent（在新 DNA 合成过程中检测氢离子）。
一般来说，边合成边测序涉及向不断增长的 DNA 链添加单核苷酸并检测每次添加。
同时，连接测序依赖于检测短 DNA 探针与片段的连接来确定序列。

```{dropdown} <i class="fa-solid fa-thumbs-up"></i>   优势
- Second-generation sequencing is often the cheapest option concerning required chemicals.
- Sparse material can still be used as input.
- High sensitivity to detect low-frequency variants and comprehensive genome coverage.
- High capacity with sample multiplexing.
- Ability to sequence thousands of genes simultaneously.
```

```{dropdown} <i class="fa-solid fa-thumbs-down"></i></i>   局限性
- The sequencing machines are expensive and often must be shared with colleagues.
- Second-generation sequencers are big, stationary machines that are not designed for fieldwork.
- Generally, second-generation sequencing results in many short sequencing fragments (reads) that are hard to use for novel genomes.
- The quality of the sequencing result depends on the reference genome.
```

### 第三代测序

第三代测序，现在也称为下一代测序，为市场带来了两项创新。
首先是长读长测序，它能够检测比二代测序产生的核苷酸片段长得多的核苷酸片段。
典型的 Illumina 短读长测序仪可生成 75 至 300 个碱基对的片段，具体取决于型号。
通过第三代测序，测序仪可以读取数万个碱基对。
这对于在没有可用参考基因​​组的情况下组装新基因组尤其重要。
其次，实时测序能力是第三代测序的另一重大进步。
与体积小且不需要更复杂的化学机器的便携式测序仪相结合，测序现在可以“现场准备”，甚至可以在远离实验室设施的地方用于收集样本。

```{admonition} A note on sequencing length
:class: note, dropdown
- 1 base pair (bp)
- 1 kilo base pair (kb) = 1,000 bp
- 1 mega base pair (Mb) = 1,000,000 bp
- 1 giga base pair (Gb) = 1,000,000,000 bp
```

Pacific Biosciences (PacBio) 于 2010 年推出了零模式波导测序，它使用包含单个 DNA 聚合酶的所谓纳米孔。
这使得任何单个核苷酸的掺入都可以通过安装在纳米孔下方的检测器直接观察到。
每种类型的核苷酸都用特定的荧光染料标记，该染料在掺入过程中会发出荧光信号，随后将其作为序列读数进行测量。
从 PacBio 测序仪获得的读数通常为 8 至 15 kb，最多可能为 70 kb。

Oxford Nanopore Technologies 于 2012 年首次推出便携式 MinION 设备。The MinION 及其后继产品 GridION 和 PromethION 产品线是 DNA/RNA 的测序仪，能够产生超过 2 Mb 的读数。
通常，纳米孔单细胞文库产生的 cDNA 转录本长度通常为 900 bp 至 2.7 kb。
值得注意的是，MinION 测序设备足够小，可以放在手掌中。
Oxford Nanopore 测序仪的想法是检测当核酸迁移通过蛋白质纳米孔 {cite}`Jain2016`时电流的变化。

```{dropdown} <i class="fa-solid fa-thumbs-up"></i>   优势
- Long reads will allow for the assembly of large novel genomes.
- Sequencers are portable, making them ideal for fieldwork.
- Possibility to directly detect epigenetic modifications of DNA and RNA sequences.
- Speed! Third-generation sequencers are fast.
```

```{dropdown} <i class="fa-solid fa-thumbs-down"></i></i>   局限性
- Some third-generation sequencers exhibit higher error rates than second-generation sequencers (Roche's new sequencing by expansion technology, for example, attempts to solve this problem {cite}`Jain2016`
{cite}`roche`).
- The reagents are generally more expensive than second-generation sequencing.
```

## NGS 流程概述

尽管存在多种 NGS 技术，但对 DNA 进行测序（因此反转录 RNA）的一般步骤基本相同。
差异主要在于各自测序技术的化学原理。

1. **样品和文库制备**：第一步，通过将 DNA 样品片段化并将其与 {term}`adapter molecules <Adapter sequences>`连接来制备所谓的 {term}`library`。
   这些接头参与文库片段与基质的杂交并形成引发位点。

2. **扩增和测序**：在第二步中，文库被转化为单链分子。
   在扩增步骤（例如聚合酶链式反应）期间，会创建 DNA 分子簇。
   所有簇在单次测序运行期间执行单独的反应。

3. **数据输出和分析**：测序实验的输出取决于测序技术和化学。
   一些测序仪生成荧光信号，这些信号存储在特定的输出文件中。
   其他的可以产生电信号，并以相应的文件格式存储。
   一般来说，生成的数据量（原始数据）是巨大的。
   此类数据需要复杂且计算量大的处理。
   这将在原始数据处理章节中进一步讨论。

## RNA 测序

到目前为止，我们仅介绍了未提及的测序假设，即 DNA 正在被测序。
然而，了解生物体的 DNA 序列及其调节元件的位置，对细胞的动态和实时过程知之甚少。
RNA 测序 (RNA-Seq) 允许科学家在测序时以基因表达谱的形式获得细胞、组织或生物体的快照。
该信息可用于检测疾病状态对治疗、环境因素、基因型和其他实验条件的反应的变化。

RNA-Seq 很大程度上遵循 DNA 测序方案，但包括一个逆转录步骤，其中 {term}`complementary DNA (cDNA)`从 RNA 模板合成。
现代 RNA 测序允许对转录本进行无偏采样，与基于微阵列的测定或定量逆转录 PCR 相比，后者需要探针设计来专门针对感兴趣的区域。
{ref}`Microarray-based assays <seealso-dropdown>`使用探针、互补序列来检测感兴趣的特定序列（例如基因）。
{ref}`Quantitative reverse transcription PCR <seealso-dropdown>`通过监测 PCR 期间互补 DNA (cDNA) 分子的扩增来测量目标 RNA 的量。

获得的基因表达谱进一步能够检测基因亚型、基因融合、单核苷酸变异和许多其他有趣的特性。
现代 RNA 测序不受先验知识的限制，并且允许捕获已知和新颖的特征。
这会产生丰富的数据集，可用于探索性数据分析。

## 单细胞RNA 测序

### 批量测序与单细胞 RNA 测序

RNA-Seq 主要可以通过两种方式进行：对来自感兴趣来源的跨细胞 ({term}`bulk sequencing <Bulk RNA sequencing>`) 的混合 RNA 进行测序，或对细胞的转录组进行单独测序（单细胞测序）。
在大多数情况下，混合所有细胞的 RNA 比实验复杂的单细胞 RNA-Seq (scRNA-seq) 更便宜、更容易
批量 RNA-Seq 产生细胞平均表达谱，通常更容易分析，但也隐藏了一些复杂性，例如细胞表达谱异质性，这可能有助于回答感兴趣的问题。
某些药物或干扰可能仅影响特定的 {term}`cell types <Cell type>`或细胞类型之间的相互作用。
例如，在肿瘤学中，可能存在导致复发的罕见耐药肿瘤细胞，即使在培养细胞上，也很难通过简单的批量 RNA-Seq 来识别。

为了揭示这种关系，在单细胞水平上检查基因表达至关重要。
然而，scRNA-seq 确实有一些警告。
首先，单细胞实验通常更昂贵且更难以正确进行。
其次，由于分辨率的提高，{term}`downstream analysis`变得更加复杂，并且更容易得出错误的结论。

单细胞实验通常遵循与批量 RNA-Seq 实验类似的步骤（见上文），但需要进行一些调整。
与批量测序一样，单细胞测序需要裂解、反转录、扩增和最终测序。
此外，单细胞测序需要细胞分离和物理分离到较小的反应室或另一种形式的细胞标记，以便能够稍后将获得的转录组映射回起源细胞。
因此，这些也是大多数单细胞测定的不同步骤：单细胞分离、转录物扩增和测序，具体取决于测序仪。
但在我们开始解释单细胞 RNA 测序的复杂性之前，有必要了解以如此高分辨率测量 {term}`mRNA <Messenger RNA (mRNA)>`时出现的生物学和技术挑战。

#### 测量“信使”

scRNA-seq 的核心在于一个基本问题：我们到底想要测量什么？
在 RNA-seq 实验中，我们的重点是量化单个细胞内的信使 RNA (mRNA)。
正如 Brenner、Jacob 和 Meselson 在 1961 年所描述的那样，该分子是“一种不稳定的中间体，它将信息从基因携带到核糖体以进行蛋白质合成”，并因此创造了术语“信使”{cite}`brenner1961unstable`。
因此，mRNA 是 DNA 和蛋白质生产之间的关键纽带。
然而，mRNA 仅占细胞总数 RNA 的一小部分。
RNA 质量的大约 3–7% 是 mRNA，而绝大多数是非编码 RNA：80–90% 核糖体 RNA (rRNA)、10–15% 转移 RNA (tRNA) 和约 1% 其他非编码物种{cite}`palazzo2015non`([overview of non-coding RNA](https://www.bio-rad.com/de-de/applications-technologies/coding-non-coding-rna?ID=Q1070M70KWE7))。
据估计，典型的哺乳动物细胞中有 100,000 至 1,000,000 个 mRNA 分子，覆盖了所有基因 {cite}`velculescu1999analysis, Islam2014`的约 50%。
这意味着在任何给定细胞中，大量基因根本不转录——这反映了细胞的特定身份和功能。
然而，当前 scRNA-seq 技术的技术限制使测量变得更加复杂。
例如，像 10X Genomics 这样的流行平台每次运行仅捕获最多 65% 的细胞，并且仅回收每个细胞的 mRNA {cite}`aljanahi2018introduction`的约 14%。
这些限制使得检测弱表达基因尤其具有挑战性。

通过这些数值镜头了解基因表达不仅揭示了生物学的复杂性，还揭示了我们工具的局限性 ({numref}`quantifying-gene-expression`)。
为了更深入地理解这一点，让我们一步步从基因到蛋白质。

#### From Gene 至蛋白质

我们的旅程从一个基因开始，它是 DNA 中的一个定义区域，充当 mRNA 合成的模板。
虽然个体之间的基因数量略有不同（约 70 个基因），但人类基因组平均包含大约 22,000 个基因 {cite}`pertea2010between`。
基因转录远非连续的。
相反，它发生在随机爆发中——短暂的、不规则的活动周期，在此期间，基因可能突然产生多个 mRNA 转录本，然后返回沉默 {cite}`suter2011mammalian`。
这也是我们使用负二项式分布对 mRNA 转录进行建模的原因。
这种分布是理想的，因为它对事件计数 (mRNA) 进行建模，同时捕获由转录爆发 {cite}`ren2020negative,love2015deseq2`引起的过度分散（方差超过平均值）。

最初的 RNA 转录本，称为前 mRNA，然后进行选择性剪接，该过程允许转录本的不同区域（称为内含子和外显子）以多种方式连接。
这意味着单个基因可以产生多个不同的 mRNA 亚型。
平均而言，每个人类基因产生约 3.4 个 mRNA 同工型 {cite}`lee2015mechanisms`。
虽然所有人类基因都至少有两种替代亚型，但有些基因突破了复杂性的极限。
例如，人类 basonuclin 2 基因有可能产生多达 90,000 种 mRNA 亚型，从而产生 2,000 多种不同的蛋白质 {cite}`vanhoutteghem2007human`。
然而，在某些情况下，选择性剪接也可能导致酶无功能并诱发疾病状态。
最后，这个“成熟的”mRNA 被翻译成蛋白质。
在这里，数字也有很大差异。
在哺乳动物中，蛋白质与 mRNA 的比例中值估计约为每 mRNA {cite}`li2014system`10,000 个蛋白质。
然而，每个转录本的蛋白质范围可以从几百到近一百万个，具体取决于基因、细胞类型和许多其他因素{cite}`edfors2016gene`。
最终，这个过程在单个人类细胞{cite}`milo2013total`内产生大约十亿个蛋白质。

了解这些层——从转录爆发和选择性剪接到蛋白质翻译——凸显了基因表达不仅仅是一个静态途径，而是一个动态的概率系统。
以单细胞分辨率测量它可以提供深刻的见解，但也揭示了我们当前技术的挑战和局限性。

### 转录本定量

转录本定量是将原始数据转换为每个样本每个基因（对于批量测序）或每个细胞（对于单细胞测序）的估计转录本计数表的过程。
有关此计算过程的更多详细信息将在 [next chapter](../introduction/raw_data_processing.md) 中描述。

转录本定量有两种主要方法：全长和基于标签。
全长方案尝试通过测序读取均匀地覆盖整个转录本，而基于标签的方案仅捕获 5' 或 3' 末端。
转录本定量方法对捕获的基因有很大影响，因此分析人员必须了解所使用的定量过程。
全长测序仅限于基于板的方案 [(see below)](#plate-based)，文库制备与批量 RNA-seq 测序方法相当。
全长方案并不总能实现转录本的均匀覆盖，因此基因体的特定区域可能仍然存在偏差。
全长协议的一个主要优点是它们允许检测剪接变体。

基于标签的方案仅对转录本的 3' 或 5' 末端进行测序。
这是以不（必然）覆盖整个基因长度为代价的，使得难以明确地将读数与转录本对齐并区分不同的亚型{cite}`Archer2016`。
然而，它允许使用独特的分子标识符 ({term}`UMIs <Unique Molecular Identifier (UMI)>`)，这对于解决转录物扩增过程中的偏差很有用。

转录本扩增过程是任何 RNA-seq 测序运行中的关键步骤，以确保转录本足够丰富以进行质量控制和测序。
在此过程中，通常通过聚合酶链式反应 ({term}`PCR`) 进行，由原始分子的相同片段生成副本。
由于副本和原始分子无法区分，因此确定样品中分子的原始数量变得具有挑战性。
UMI 是量化原始非重复分子的常见解决方案。

UMI 充当分子 {term}`barcodes <barcode>`，有时称为随机条形码。
这些“条形码”由短的随机核苷酸序列组成，作为独特的标签添加到样品中的每个分子中。
必须在扩增步骤之前的文库生成过程中添加 UMI。
准确识别 PCR 重复项对于下游分析排除或了解 {term}`amplification biases <amplification bias>`{cite}`Aird2011`非常重要。

扩增偏差是优先扩增的 RNA/cDNA 序列的术语，因此会更频繁地进行测序，从而导致更高的计数。
它可能会损害任何基因表达分析，因为不太活跃的基因可能突然表现出高度表达。
对于在 PCR 步骤的后期阶段扩增的序列尤其如此，其中错误率可能已经比早期的 PCR 阶段相对较高。
尽管在计算上可以通过过滤掉具有相同比对坐标的读数来检测和删除此类序列，但通常建议如果可能的话，始终使用 UMI 设计实验。
Using UMIs 进一步允许在不损失准确性的情况下标准化基因计数 {cite}`Kivioja2012`。

### 单细胞测序方案

存在许多用于对单个细胞的转录组进行测序的方案。
然而，这些术语通常可能含糊不清，尤其是对于该领域的新手来说。
为了澄清这一点，我们根据细胞的分离方式将这些技术分为两大类：

- **液滴分离**：这些方法将单个细胞封装到乳液内的微小液滴中，从而实现高通量处理。
- **物理隔室中的分离**：这些技术将细胞分离到不同的物理隔室中，通常称为孔。

每种方法在恢复转录本的能力、测序细胞的数量以及许多其他方面都有所不同。
在以下小节中，我们将简要讨论它们的工作原理、优点和缺点，以及数据分析师应注意的有关各自协议的可能偏见。

#### 液滴分离

##### 最常见的协议

最广泛使用的协议是 **inDrop** {cite}`Klein2015`、**Drop-seq** {cite}`exp:Macosko2015`和市售的 **10x Genomics Chromium** {cite}`exp:Zheng2017`。
这些方案使用微流体将细胞捕获在称为液滴的微小水泡中。
每个液滴形成一个单独的空间，仅包含一个细胞和所需的化学物质（珠子）。
上述协议每秒可以生成数千次液滴。
这种大规模并行过程以相对较低的成本产生大量液滴。

```{admonition} Droplets by vortexing
:class: dropdown, note
The PIP-seq protocol offers a simplified alternative to traditional microfluidic methods for generating monodispersed water-in-oil droplets.
Unlike complex microfluidic devices that require specialized equipment and expertise, PIP-seq achieves droplet formation through simple vortexing of the solution.
This method can be easily scaled by increasing the container volume without being constrained by emulsion time, a common limitation of microfluidics {cite}`clark2023microfluidics`.

However, despite its simplicity, independent {term}`benchmarks <benchmark>` indicate that PIP-seq still has limitations compared to well-established methods.
For instance, PIP-seq achieved approximately 1,500 gene counts, whereas the best 10x Genomics Chromium kit showed around 4,000 gene counts {cite}`de2025comprehensive`.
These findings highlight a trade-off between ease of use and performance in the current version of the PIP-seq protocol.
```

尽管这三种方案在细节上有所不同，但含有封装细胞的纳升大小的液滴始终被设计为同时捕获珠子和细胞。
封装过程使用专门的微珠进行，珠上引物包含 PCR 手柄、细胞条形码、4-8b 碱基对长 UMI 和聚 T 尾（或者在 5' 试剂盒的情况下，将有聚 T 引物）。
裂解后，细胞的 mRNA 立即释放并被附着在珠子上的带条形码的寡核苷酸捕获。
接下来，收集并破碎液滴以释放附着在微粒上的单细胞转录组（STAMP）。
接下来是 PCR 和逆转录以捕获和扩增转录本。
最后，进行标记，随机切割转录本并连接测序接头。
如上所述，该过程产生可供测序的测序文库。
在基于液滴的方案中，仅回收了约 10% 的细胞转录本 {cite}`Islam2014`。
值得注意的是，这种低测序足以对细胞类型进行可靠的鉴定。

所有三种方法都会导致特征偏差。
不同协议所使用的珠子材料有所不同。
{term}`Drop-seq`使用脆性树脂作为珠子。
因此，珠子采用 {term}`Poisson distribution`封装，而 {term}`InDrop`和 10X Genomics 珠子可变形，导致珠子占有率超过 80% {cite}`Zhang2019`。

此外，捕获效率可能受到 Drop-Seq 中表面束缚引物使用的影响。
InDrop 使用通过光裂解释放的引物，10X 基因组溶解珠子。
这种差异也会影响逆转录过程的位置。
In Drop-seq，逆转录发生在珠子从液滴中释放后，而对于 InDrop 和 10X 基因组协议 {cite}`Zhang2019`来说，逆转录发生在液滴内部。

张等人的比较2019 年发现 inDrop 和 Drop-seq 在珠子质量方面优于 10X Genomics，因为前两个系统中的细胞条形码包含明显的不匹配。
此外，10X Genomics 中来自有效条形码的读取比例为 75%，而 InDrop 中仅为 25%，Drop-seq 中仅为 30%。

10X Genomics 也证明了在灵敏度方面的类似优势。
在比较过程中，10X Genomics 平均捕获了来自 3000 个基因的约 17000 个转录本，而 Drop-seq 捕获了 2500 个基因的 8000 个转录本，InDrop 捕获了 1250 个基因的 2700 个转录本。
10X Genomics 的技术噪音最低，其次是 Drop-seq 和 InDrop {cite}`Zhang2019`。

实际生成的数据显示出很大的协议偏差。
10X Genomics 倾向于捕获和扩增较短的基因和具有较高 GC 含量的基因，而相比之下，Drop-seq 更喜欢具有较低 GC 含量的基因。
尽管 10X Genomics 在各个方面均优于其他方案，但每个细胞的成本也约为其他方案的两倍。
此外，除了珠子之外，Drop-seq 是开源的，如果需要，可以更轻松地调整协议。
InDrop 是完全开源的，甚至连珠子都可以在实验室中制造和修改。
因此，InDrop 是三种协议中最灵活的。

```{dropdown} <i class="fa-solid fa-thumbs-up"></i>   优势
- Allows for the cost-efficient sequencing of cells in large quantities to identify the overall composition of a tissue and characterize rare cell types.
- UMIs can be incorporated.
```

```{dropdown} <i class="fa-solid fa-thumbs-down"></i></i>   局限性
- Low detection rates of transcripts compared to other methods.
- Captures only 3' ends (or 5' ends, depending on kit) and not full transcripts.
```

##### 纳米孔测序遇上液滴技术

长读长单细胞测序方法很少使用 UMI {cite}`Singh2019`或不执行 UMI 校正 {cite}`Gupta2018`，因此将一些读段错误分配给新的 UMI。
由于长读长测序仪的测序错误率较高，这会导致严重的问题{cite}`Lebrigand2020`。
勒布里甘等人。推出了 ScNaUmi-seq（使用 UMI 的单细胞纳米孔测序），将纳米孔测序与细胞条形码和 UMI 分配相结合。
通过将 Nanopore 读数中发现的细胞条形码序列与从相同区域或基因 {cite}`Lebrigand2020`的 Illumina 读数中恢复的细胞条形码序列进行比较，以 Illumina 数据指导条形码分配。
然而，这实际上需要两个单细胞库。
ScCOLOR-seq 使用在条形码全长范围内互补的核苷酸对，通过计算方式准确无误地识别条形码。
然后，这些条形码将用作纠正剩余错误条形码 {cite}`Philpott2021`的指南。
修改后的 UMI 工具基于定向网络的方法可纠正 UMI 序列重复。

```{dropdown} <i class="fa-solid fa-thumbs-up"></i>   优势
- Recovers splicing and sequence heterogeneity information
```

```{dropdown} <i class="fa-solid fa-thumbs-down"></i></i>   局限性
- Nanopore reagents are expensive.
- High cell barcode recovery error rates.
- Depending on the protocol, barcode assignment is guided with Illumina data requiring two sequencing assays.
```

##### 基于板的协议

通常，基于板的方案将细胞物理分离到微孔板中。
第一步需要通过例如荧光激活细胞分选（FACS）进行细胞分选，其中根据特定的细胞表面标记对细胞进行分选；或通过微量移液。
然后将选定的细胞放入含有细胞裂解缓冲液的各个孔中。
然后在这些孔中进行逆转录。
这样可以在一次实验中分析数百个细胞，每个细胞捕获 5000 至 10000 个基因。

基于板的测序方案包括但不限于 SMART-seq2、MARS-seq、QUARTZ-seq 和 SRCB-seq。
一般来说，这些协议的不同在于其复用能力。
例如，MARS-seq 允许三个条形码级别，即分子、细胞和板级标签，以实现强大的复用功能。
相反，SMART-seq2 不允许早期复用，从而限制了细胞数量。
Mereu 等人对协议进行了系统比较。 2020 年的研究表明，QUARTZ-seq2 可以比 SMART-seq2、MARS-seq 或 SRCB-seq 每个细胞 {cite}`Mereu2020`捕获更多的基因。
这意味着 QUARTZ-seq2 可以很好地捕获细胞类型特异性标记基因，从而实现可靠的细胞类型注释。

```{dropdown} <i class="fa-solid fa-thumbs-up"></i>   优势
- Recovers many genes per cell, allowing for a deep characterization.
- It is possible to gather information before the library preparation, e.g., through FACS sorting to associate information such as cell size and the intensity of any used labels with good coordinates.
- Allows for full-length transcript recovery.
```

```{dropdown} <i class="fa-solid fa-thumbs-down"></i></i>   局限性
- The scale of plate-based experiments is limited by the lower throughput of their individual processing units.
- Fragmentation step eliminates strand-specific information {cite}`Hrdlickova2017`.
- Depending on the protocol, plate-based protocols might be labor-intensive with many required pipetting steps, leading to potential technical noise and {term}`batch effects <batch effect>`.
```

##### Fluidigm C1

商用 Fluidigm C1 系统是一种微流控芯片，能够以自动化方式将细胞装载并分离到小型反应室中。
The CEL-seq2 和 SMART-seq（版本 1）协议在其工作流程中使用 Fluidigm C1 芯片，允许同时进行 RNA 提取和文库制备步骤，从而减少所需的体力劳动。
然而，Fluidigm C1 需要相当均匀的细胞混合物，因为细胞将根据其大小到达微流控芯片上的不同位置，这可能会引入潜在的位置偏差。
由于扩增步骤在各个孔中进行，因此可以进行全长测序，从而有效减少许多其他单细胞 RNA-seq 测序方案的 3' 偏差。
该方案通常也更昂贵，因此主要用于特定细胞群的广泛检查。

```{dropdown} <i class="fa-solid fa-thumbs-up"></i>   优势
- Allows for full-length transcript coverage.
- Splicing variants and T/B cell receptor repertoire diversity can be recovered.
```

```{dropdown} <i class="fa-solid fa-thumbs-down"></i></i>   局限性
- Only allows for the sequencing of up to 800 cells {cite}`fluidigm`.
- More expensive per cell than other protocols.
- Only about 10% of the extracted cells are captured, which makes this protocol unsuitable for rare cell types or low input.
- The used arrays only capture specific cell sizes, which may bias the captured transcripts.
```

#### 概括

总之，我们强烈建议湿实验室和干实验室科学家根据研究目的选择测序方案。
是否需要对特定细胞类型群体进行深入表征？
在这种情况下，基于板的方法之一可能更合适。
相反，基于液滴的测定将更好地捕获异质混合物，从而对测序细胞进行更广泛的表征。
此外，如果预算是一个限制因素，那么选择的协议应该更具成本效益和稳健。
分析数据时，请注意测序分析特定的偏差。
为了对所有单细胞测序方案进行广泛比较，我们推荐 Mereu 等人撰写的“细胞图谱项目的单细胞 RNA 测序方案基准测试”论文。 {cite}`Mereu2020`。


### 单细胞与单核

到目前为止，我们只讨论了单细胞测定，但也可以仅对细胞核进行测序。
单细胞分析并不总是提供特定组织或器官（例如大脑）的细胞类型的公正视图。
在组织解离过程中，某些细胞类型更脆弱，因此难以捕获。例如，在小鼠新皮质 {cite}`Tasic2018`中观察到快速尖峰小白蛋白阳性中间神经元和皮质下突出的谷氨酸能神经元的比例低于预期。
相反，非神经元细胞比神经元更好地在解离后存活，并且在成人新皮质 {cite}`darmanis2015`的单细胞悬浮液中过多。
此外，单细胞测序高度依赖新鲜组织，因此很难使用组织生物库。
另一方面，细胞核对机械力的抵抗力更强，可以安全地从冷冻组织中分离出来，而无需使用组织解离酶 {cite}`Krishnaswami2016`。
这两种选择在不同的组织和样本类型中都有不同的适用性，并且由此产生的偏差和不确定性尚未完全揭示。
已经表明，细胞核准确地反映了细胞 {cite}`Ding2020`的所有转录模式。
实验设计中单细胞与单核的选择主要取决于组织样本的类型。
然而，数据分析应该意识到解离能力将对潜在可观察的细胞类型产生强烈影响。
因此，我们强烈鼓励湿实验室和干实验室科学家之间就实验设计进行讨论。
